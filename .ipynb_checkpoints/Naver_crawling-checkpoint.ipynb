1. 목표 = 미션 : 네이버 포털사이트 내 뉴스 카테고리에서 특정 키워드 검색 시, 나오는 기사 제목을 수집하기
2. 그동안 배웠던 문법 즉, 함수가 먼저 나오면 안 됨
3. 위 목표 및 미션을 해결하기 위한 절차 & 과정 = 프로세스를 먼저 그려야 함

컴퓨터 입장
1. 네이버 포털 사이트 > 뉴스 카테고리
2. 검색을 희망하는 키워드를 입력
3. 키워드 입력 후 검색하기 버튼 클릭
4. 버튼 클릭 된 후 출력되고 있는 기사 제목 스캔
5. 기사 제목 스캔 후 상위 10개 기사 제목만 가져와야 함

```python
import requests
from bs4 import BeautifulSoup

query = input("검색하고 싶은 키워드를 입력하세요")
query = "디지털마케팅"
url = f"https://search.naver.com/search.naver?where=news&ie=utf8&sm=nws_hty&query={query}"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.0.0 Safari/537.36"
}

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, "html.parser")

titles = soup.select("a.fender-ui_228e3bd1.ise_N8_wf_ooMaN2")
# soup이라는 함수 안에서 select라는 메서드를 쓰기 위해선 인자값과 동일한 값이기만 하면 몇 개든지 찾아와

for i, title in enumerate(titles[:10], 1) :
    print(f"{i}. {title.text}")
```
