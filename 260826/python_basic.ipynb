1. 프로그래밍 언어의 역사
- 절차지향 프로그래밍 언어 (c언어 : 대부분의 프로그래밍 언어의 조상격)
- 절차지향 > 컴퓨터가 이해하기 편하도록 > 모든 실행값 문법의 절차에 반드시 맞춰서 작성
- 객체지향 프로그래밍 언어 (Python, Javascript, Java) -> 다들 자신만의 class를 가지고 있음 -> class에 넣으면 instance가 나옴
- 절차지향의 좋은 점만 수용 / 융통성이 없는 문법은 배제
- 절차를 지키지 않으면서 -> 클래스 (class -> 붕어빵 틀) 반죽은 집에서 만들고, 모양 틀 기계에 넣으면 100명, 1000명도 처리 가능

```python
# 2) 정수라는 값이 형태의 값이 변수에 할당된다면, 가장 많이 사용하는 문법 -> 연산자
# 사칙 연산 = +, - ,*, / + @ (나머지 연산, 제곱근, 몫)

intA = 7
intB = 3

print(intA + intB)
print(intA - intB)
print(intA * intB)
print(intA / intB)
```

```python
print(intA // intB) # 7 / 3 의 몫
print(intA % intB) # 7 % 3 의 나머지 -> 값이 짝수인지, 홀수인지
print(intA ** intB) # 7^3 
```

```python
# flaot > int 타입에 사용가능한 모든 연산자 다 사용가능!
# boolean > True, False > 명제 > 조건문, 반복문

print(intA > intB) # 비교연산
print(intA < intB) 
print(intA == intB) # = -> 변수 선언, 할당에 쓰임  == -> 같다에 쓰임
print(intA != intB) # ! => 부정연산자 = not
print(intA >= intB) 
print(intA <= intB) 
```
