```python
# type() : 함수
# count() : 매서드
# 함수는 앞에 어떤 것의 도움을 받지 않고 독립적으로 사용가능!
# 매서드는 사전적 의미 : 방법 -> 대부분 앞의 어떤 것(=객체)의 도움을 받아야 함

article.count("korea") #인자값으로 입력된 문자열이 몇 번 사용되었는가?
print(article.count("Korea")) #대소문자 구분함
print(len(article)) #전체 문자열 길이 개수 (빈칸, 특수문자 모두 포함)
print(article.find("S")) # 전체 문자열 중 몇 번째 인덱스 값으로 해당 인자값이 사용되었는지 확인할 때!
```
